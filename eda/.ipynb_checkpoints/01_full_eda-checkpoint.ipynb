# likable-river — Full-data EDA (all 14 parquets)
Team likable-river, PRC Data Challenge 2026. Target: DEP taxi-out seconds, RMSE.
This notebook scans the FULL data (no sampling) and ends with programmatic assertions.

```python
import glob, os
import numpy as np
import pandas as pd
import pyarrow.parquet as pq
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt

DATA = '../../data'
FIG = 'figures'
os.makedirs(FIG, exist_ok=True)
TRAIN = sorted(glob.glob(f'{DATA}/training_*.parquet'))
RANK = f'{DATA}/ranking.parquet'
SUB = f'{DATA}/submitting.parquet'
print(len(TRAIN), 'train files')
pd.set_option('display.width', 180); pd.set_option('display.max_columns', 40)
```

## 1. Manifest — rows, sizes, schema identity

```python
rows = []
sch = None
for f in TRAIN + [RANK, SUB]:
    pf = pq.ParquetFile(f)
    cols = pf.schema_arrow.names
    if sch is None:
        sch = cols
    rows.append({'file': os.path.basename(f), 'rows': pf.metadata.num_rows,
                 'cols': len(cols), 'MB': round(os.path.getsize(f)/2**20, 1),
                 'same_cols': cols == sch})
man = pd.DataFrame(rows)
print(man.to_string())
print('train total:', man.iloc[:12]['rows'].sum())
assert man['same_cols'][:13].all(), 'schema drift in 30-col files'
```

## 2. IDs — uniqueness, disjoint ranges, template 1:1

```python
seen = []
for f in TRAIN + [RANK]:
    t = pq.read_table(f, columns=['MVT_ID_mvt']).to_pandas()
    v = t['MVT_ID_mvt']
    assert v.isna().sum() == 0 and v.nunique() == len(v), f'dup/null in {f}'
    seen.append((v.min(), v.max()))
    print(os.path.basename(f), len(v), int(v.min()), int(v.max()))
iv = sorted(seen)
print('NOTE: ranges need not be disjoint (Oct/Nov interleave) — enforcing global uniqueness instead')
all_ids = pd.concat([pq.read_table(f, columns=['MVT_ID_mvt']).to_pandas() for f in TRAIN + [RANK]], ignore_index=True)['MVT_ID_mvt']
print('all files total:', len(all_ids), 'unique:', all_ids.nunique())
assert all_ids.nunique() == len(all_ids), 'cross-file duplicate IDs!'
print('IDs unique within + across files: OK')
rk = pq.read_table(RANK, columns=['MVT_ID_mvt', 'PHASE_mvt']).to_pandas()
sb = pq.read_table(SUB).to_pandas()
dep_ids = rk.loc[rk.PHASE_mvt == 'DEP', 'MVT_ID_mvt']
print('sub shape:', sb.shape, 'taxi all-null:', sb['TAXITIME_SEC_mvt'].isna().all())
assert len(sb) == len(dep_ids) == 344841
assert set(sb['MVT_ID_mvt']) == set(dep_ids), 'template IDs != ranking DEP'
assert (sb['MVT_ID_mvt'].values == dep_ids.values).all(), 'order differs'
print('template 1:1 + order-identical: OK')
```

## 3. Airports — the 10-vs-11 verdict, counts

```python
ap = []
for f in TRAIN + [RANK]:
    t = pq.read_table(f, columns=['ADEP_mvt', 'ADES_mvt', 'PHASE_mvt']).to_pandas()
    d = sorted(t.loc[t.PHASE_mvt == 'DEP', 'ADEP_mvt'].unique())
    a = sorted(t.loc[t.PHASE_mvt == 'ARR', 'ADES_mvt'].unique())
    ap.append((os.path.basename(f), d, a))
    assert len(d) == len(a) == 10, f
base = ap[0][1]
assert all(d == base and a == base for _, d, a in ap)
print('ALL 13 files: exactly these 10:', base)
full = pd.concat([pq.read_table(f, columns=['ADEP_mvt', 'PHASE_mvt']).to_pandas() for f in TRAIN], ignore_index=True)
vc = full.loc[full.PHASE_mvt == 'DEP', 'ADEP_mvt'].value_counts()
print(vc.to_string())
print((vc / vc.sum()).round(4).to_string())
```

## 4. Target identity + distribution + outliers (FULL train DEP)

```python
dep = pd.concat([pq.read_table(f, columns=['MVT_TIME_UTC_mvt', 'BLOCK_TIME_UTC_mvt',
    'TAXITIME_SEC_mvt', 'PHASE_mvt', 'ADEP_mvt']).to_pandas() for f in TRAIN], ignore_index=True)
dep = dep[dep.PHASE_mvt == 'DEP'].reset_index(drop=True)
print('train DEP:', len(dep))
d = (dep['MVT_TIME_UTC_mvt'] - dep['BLOCK_TIME_UTC_mvt']).dt.total_seconds()
mm = (d != dep['TAXITIME_SEC_mvt'])
print('identity mismatches:', mm.sum(), '/', len(dep))
assert mm.sum() == 0
y = dep['TAXITIME_SEC_mvt']
print(y.describe(percentiles=[.5, .75, .9, .95, .99, .999]).to_string())
print('neg:', (y < 0).sum(), 'zero:', (y == 0).sum(), '<60s:', (y < 60).sum(),
      '>30min:', (y > 1800).sum(), '>60min:', (y > 3600).sum(), '>2h:', (y > 7200).sum())
print('neg values:', sorted(y[y < 0].unique()))
print('top extremes:', sorted(y.nlargest(10).tolist()))
```

```python
g = dep.groupby('ADEP_mvt')['TAXITIME_SEC_mvt'].agg(['count', 'mean', 'median', 'std', 'min', 'max',
    lambda s: s.quantile(.9), lambda s: s.quantile(.99)])
g.columns = ['n', 'mean', 'median', 'std', 'min', 'max', 'p90', 'p99']
print(g.round(1).sort_values('median').to_string())
fig, ax = plt.subplots(figsize=(10, 5))
dep['TAXITIME_SEC_mvt'].clip(upper=4000).hist(bins=100, ax=ax)
ax.set_title('DEP taxi-out (clipped 4000s, full train)'); ax.set_xlabel('sec')
fig.savefig(f'{FIG}/dep_hist.png', dpi=80); plt.close(fig)
fig, ax = plt.subplots(figsize=(10, 5))
dep.groupby('ADEP_mvt')['TAXITIME_SEC_mvt'].median().sort_values().plot.barh(ax=ax)
ax.set_title('Median taxi-out by airport')
fig.savefig(f'{FIG}/dep_median_by_airport.png', dpi=80); plt.close(fig)
print('saved dep_hist, dep_median_by_airport')
```

## 5. Seasonality + hour/dow effects

```python
dep['blk_month'] = dep['BLOCK_TIME_UTC_mvt'].dt.month
dep['blk_hour'] = dep['BLOCK_TIME_UTC_mvt'].dt.hour
dep['blk_dow'] = dep['BLOCK_TIME_UTC_mvt'].dt.dayofweek
print(dep.groupby('blk_month')['TAXITIME_SEC_mvt'].agg(['count', 'mean', 'median',
      lambda s: s.quantile(.99)]).round(1).to_string())
h = dep.groupby('blk_hour')['TAXITIME_SEC_mvt'].median()
w = dep.groupby('blk_dow')['TAXITIME_SEC_mvt'].median()
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
h.plot(ax=axes[0], marker='o'); axes[0].set_title('Median taxi by UTC hour'); axes[0].set_xlabel('hour')
w.plot(ax=axes[1], marker='o'); axes[1].set_title('Median taxi by dow (0=Mon)'); axes[1].set_xlabel('dow')
fig.savefig(f'{FIG}/time_effects.png', dpi=80); plt.close(fig)
print('hour range:', round(h.max() - h.min(), 1), 'dow range:', round(w.max() - w.min(), 1))
print('saved time_effects')
```

## 6. ARR contrast + ranking month coverage

```python
arr = pd.concat([pq.read_table(f, columns=['MVT_TIME_UTC_mvt', 'BLOCK_TIME_UTC_mvt',
    'TAXITIME_SEC_mvt', 'PHASE_mvt']).to_pandas() for f in TRAIN], ignore_index=True)
arr = arr[arr.PHASE_mvt == 'ARR'].reset_index(drop=True)
d2 = (arr['BLOCK_TIME_UTC_mvt'] - arr['MVT_TIME_UTC_mvt']).dt.total_seconds()
print('ARR rows:', len(arr), 'BLOCK-MVT identity mismatches:', (d2 != arr['TAXITIME_SEC_mvt']).sum())
print(arr['TAXITIME_SEC_mvt'].describe(percentiles=[.5, .9, .99]).to_string())
r = pq.read_table(RANK, columns=['MVT_TIME_UTC_mvt', 'PHASE_mvt', 'BLOCK_TIME_UTC_mvt', 'TAXITIME_SEC_mvt']).to_pandas()
r['ym'] = r['MVT_TIME_UTC_mvt'].dt.strftime('%Y-%m')
print(r.groupby(['ym', 'PHASE_mvt']).size().to_string())
rd = r[r.PHASE_mvt == 'DEP']
print('ranking DEP BLOCK null:', rd['BLOCK_TIME_UTC_mvt'].isna().mean(),
      'TAXI null:', rd['TAXITIME_SEC_mvt'].isna().mean(),
      'MVT null:', rd['MVT_TIME_UTC_mvt'].isna().mean())
```

## 7. Leakage audit — AOBT_3 vs BLOCK, proxy RMSE, null map

```python
cols = ['BLOCK_TIME_UTC_mvt', 'MVT_TIME_UTC_mvt', 'TAXITIME_SEC_mvt', 'AOBT_3_flt',
        'ARVT_3_flt', 'IOBT_flt', 'EOBT_1_flt', 'LOBT_flt', 'SCHED_TIME_UTC_mvt', 'PHASE_mvt']
tr = pd.concat([pq.read_table(f, columns=cols).to_pandas() for f in TRAIN], ignore_index=True)
rk = pq.read_table(RANK, columns=cols).to_pandas()
for name, df in [('train', tr), ('ranking', rk)]:
    for ph in ['DEP', 'ARR']:
        s = df[df.PHASE_mvt == ph]
        print(name, ph, {c: round(s[c].isna().mean(), 4) for c in cols if c != 'PHASE_mvt'})
td = tr[tr.PHASE_mvt == 'DEP'].dropna(subset=['BLOCK_TIME_UTC_mvt', 'AOBT_3_flt'])
print('pairs:', len(td))
err = (td['BLOCK_TIME_UTC_mvt'] - td['AOBT_3_flt']).dt.total_seconds()
print('BLOCK-AOBT median/medAbs:', err.median(), err.abs().median())
for t in [60, 120, 300]:
    print(f'within {t}s:', round((err.abs() <= t).mean(), 4))
proxy = (td['MVT_TIME_UTC_mvt'] - td['AOBT_3_flt']).dt.total_seconds()
rmse = float(np.sqrt(np.mean((proxy - td['TAXITIME_SEC_mvt']) ** 2)))
print('MVT-AOBT proxy RMSE:', round(rmse, 1), 'corr:', round(proxy.corr(td['TAXITIME_SEC_mvt']), 4))
print('AOBT minute-precision:', bool(((td['AOBT_3_flt'].dt.second == 0)).all()))
```

## 8. Join quality + categoricals + novelty vs ranking

```python
jc = ['FLIGHT_ID_mvt', 'CALLSIGN_flt', 'ADEP_mvt', 'ADEP_flt', 'ADES_mvt', 'ADES_flt',
      'AIRCRAFT_TYPE_mvt', 'AIRCRAFT_TYPE_flt', 'FLIGHT_RULE_mvt', 'FLIGHT_RULE_flt']
tj = pd.concat([pq.read_table(f, columns=jc + ['PHASE_mvt']).to_pandas() for f in TRAIN], ignore_index=True)
rj = pq.read_table(RANK, columns=jc + ['PHASE_mvt']).to_pandas()
print('train join-fail:', round(tj['CALLSIGN_flt'].isna().mean(), 4),
      'DEP:', round(tj[tj.PHASE_mvt == 'DEP']['CALLSIGN_flt'].isna().mean(), 4))
print('rank join-fail:', round(rj['CALLSIGN_flt'].isna().mean(), 4),
      'DEP:', round(rj[rj.PHASE_mvt == 'DEP']['CALLSIGN_flt'].isna().mean(), 4))
for a, b, n in [('ADEP_mvt', 'ADEP_flt', 'ADEP'), ('ADES_mvt', 'ADES_flt', 'ADES'),
                ('AIRCRAFT_TYPE_mvt', 'AIRCRAFT_TYPE_flt', 'TYPE')]:
    both = tj.dropna(subset=[a, b])
    print(n, 'agree:', round((both[a] == both[b]).mean(), 4))
full_cat = pd.concat([pq.read_table(f, columns=['RUNWAY_mvt', 'STAND_mvt', 'AIRCRAFT_TYPE_mvt',
    'AIRCRAFT_OPERATOR_flt', 'ADES_mvt', 'MARKET_SEGMENT_flt', 'WK_TBL_CAT_flt', 'FLIGHT_TYPE_flt',
    'PHASE_mvt']).to_pandas() for f in TRAIN], ignore_index=True)
rn_cat = pq.read_table(RANK, columns=['RUNWAY_mvt', 'STAND_mvt', 'AIRCRAFT_TYPE_mvt',
    'AIRCRAFT_OPERATOR_flt', 'ADES_mvt', 'PHASE_mvt']).to_pandas()
rd_cat = rn_cat[rn_cat.PHASE_mvt == 'DEP']
td_cat = full_cat[full_cat.PHASE_mvt == 'DEP']
for c in ['RUNWAY_mvt', 'STAND_mvt', 'AIRCRAFT_TYPE_mvt', 'AIRCRAFT_OPERATOR_flt', 'ADES_mvt']:
    novel = set(rd_cat[c].dropna().unique()) - set(td_cat[c].dropna().unique())
    print(c, 'train:', td_cat[c].nunique(), 'rankDEP:', rd_cat[c].nunique(), 'novel:', len(novel))
cs = tj['CALLSIGN_flt'].value_counts()
print('callsign nunq/share:', tj['CALLSIGN_flt'].nunique(), round(tj['CALLSIGN_flt'].nunique() / len(tj), 4),
      'mean rows/cs:', round(cs.mean(), 1), 'max:', cs.max())
```

## 9. Train-vs-ranking shift (permissive-policy guard)

```python
trp = tr[tr.PHASE_mvt == 'DEP'].dropna(subset=['AOBT_3_flt']).copy()
rkp = rk[rk.PHASE_mvt == 'DEP'].dropna(subset=['AOBT_3_flt']).copy()
trp['m_a'] = (trp['MVT_TIME_UTC_mvt'] - trp['AOBT_3_flt']).dt.total_seconds()
rkp['m_a'] = (rkp['MVT_TIME_UTC_mvt'] - rkp['AOBT_3_flt']).dt.total_seconds()
print('MVT-AOBT train med/mean:', trp['m_a'].median(), round(trp['m_a'].mean(), 1))
print('MVT-AOBT rank  med/mean:', rkp['m_a'].median(), round(rkp['m_a'].mean(), 1))
trp['ms'] = (trp['MVT_TIME_UTC_mvt'] - trp['SCHED_TIME_UTC_mvt']).dt.total_seconds()
rkp['ms'] = (rkp['MVT_TIME_UTC_mvt'] - rkp['SCHED_TIME_UTC_mvt']).dt.total_seconds()
print('MVT-SCHED train med:', trp['ms'].median(), 'rank med:', rkp['ms'].median())
fig, ax = plt.subplots(figsize=(10, 4))
ax.hist(trp['m_a'].clip(0, 4000), bins=100, alpha=.5, label='train', density=True)
ax.hist(rkp['m_a'].clip(0, 4000), bins=100, alpha=.5, label='ranking', density=True)
ax.legend(); ax.set_title('MVT-AOBT train vs ranking')
fig.savefig(f'{FIG}/shift_mvt_aobt.png', dpi=80); plt.close(fig); print('saved shift_mvt_aobt')
```

## 9b. Loop-4 deep checks — AOBT precision, month-matched shift, outlier attribution

```python
sec = td['AOBT_3_flt'].dt.second.fillna(-1).astype(int)
print('AOBT seconds value_counts (top):')
print(sec.value_counts().head(8).to_string())
print('share :00s:', round((sec == 0).mean(), 4))
trp['blk_month'] = trp['MVT_TIME_UTC_mvt'].dt.month
rkp['blk_month'] = rkp['MVT_TIME_UTC_mvt'].dt.month
for m, nm in [(1, 'Jan'), (7, 'Jul')]:
    a = trp[trp['blk_month'] == m]['m_a'].median()
    b = rkp[rkp['blk_month'] == m]['m_a'].median()
    c = trp[trp['blk_month'] == m]['ms'].median()
    d = rkp[rkp['blk_month'] == m]['ms'].median()
    print(f'{nm}: MVT-AOBT train25={a:.0f} rank26={b:.0f} (d={b - a:+.0f}) | MVT-SCHED train25={c:.0f} rank26={d:.0f} (d={d - c:+.0f})')
print('airport share train vs rankDEP:')
tr_ad = dep['ADEP_mvt'].value_counts(normalize=True)
rk_ad = pq.read_table(RANK, columns=['ADEP_mvt', 'PHASE_mvt']).to_pandas()
rk_ad = rk_ad[rk_ad.PHASE_mvt == 'DEP']['ADEP_mvt'].value_counts(normalize=True)
print(pd.DataFrame({'train': tr_ad, 'rank': rk_ad}).round(4).to_string())
print('>2h outliers by airport:')
print(dep[dep['TAXITIME_SEC_mvt'] > 7200].groupby('ADEP_mvt').size().sort_values(ascending=False).to_string())
print('negatives by airport:')
print(dep[dep['TAXITIME_SEC_mvt'] < 0].groupby('ADEP_mvt').size().to_string())
nullrow = tr[(tr['BLOCK_TIME_UTC_mvt'].isna()) | (tr['TAXITIME_SEC_mvt'].isna())]
print('train null-BLOCK/TAXI rows:', len(nullrow), nullrow[['PHASE_mvt']].value_counts().to_string())
```
